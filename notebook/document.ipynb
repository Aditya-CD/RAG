```python
# Create a text file
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_texts = {
    "../data/text_files/python.txt": """Python is a high-level, general-purpose programming language known for its simple and readable syntax.

It is widely used in web development, data science, artificial intelligence, machine learning, automation, and software development.

One of the biggest advantages of Python is that its syntax is easy to understand, which makes it a great language for beginners.

For example, printing something in Python is as simple as:
print("Hello World")

In simple words, Python is a programming language that lets you give instructions to a computer in a simple and readable way.""",

    "../data/text_files/fastapi.txt": """FastAPI is a modern, fast, and lightweight web framework for building APIs with Python.

It is built on top of Starlette for web functionality and Pydantic for data validation. FastAPI uses Python type hints to automatically validate incoming data and generate interactive API documentation.

One of the main advantages of FastAPI is its high performance. It supports asynchronous programming using async and await, making it suitable for applications that need to handle many requests efficiently.

FastAPI automatically generates API documentation using OpenAPI. The documentation can be accessed through interactive interfaces such as Swagger UI and ReDoc.

FastAPI is commonly used for building REST APIs, machine learning APIs, microservices, backend services, and applications that need to communicate with frontend clients.

For example, a simple FastAPI application can define an endpoint that returns a message when a user sends a request.

In simple words, FastAPI allows developers to quickly build high-performance APIs using Python with automatic validation and documentation."""
}

for filepath, content in sample_texts.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)

print("Sample text files created successfully.")
```

```python
# Text Loader
from langchain_community.document_loaders import TextLoader


loader = TextLoader("../data/text_files/python.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
# Directory Loader
from langchain_community.document_loaders import DirectoryLoader

directory_loader = DirectoryLoader(
    "../data/text_files", 
    glob="*.txt", 
    loader_cls=TextLoader, 
    loader_kwargs={"encoding": "utf-8"}
)
documents = directory_loader.load()
documents
```

```python
for content in documents:
    print(f"Document {documents.index(content)+1}: {content.page_content}")
    print(f"Metadata: {content.metadata}")
    print("\n")
```
